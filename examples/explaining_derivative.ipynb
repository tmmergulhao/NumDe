# 1st example - Toy model

For the first example, let's take derivatives of the following function:

$$
f(x,a,b)=\sin(x\,a)\cos(x\,b)\;,
$$

with respect to $a$ and $b$. The function is defined in a .py called my_functions.py. NumDe is compatible with multiprocessing, and to use it you must define a function in a separate file and import it.

```python
x_array = np.arange(-10,10,0.01)
def function_test(theta,kwargs):
    return np.sin(kwargs["x_array"]*theta[0]) * np.cos(kwargs["x_array"]*theta[1])
```

We can inizialize the FiniteDerivative method in the following way:

```python
ND_obj = nd.FiniteDerivative()
```

```python
N_taylor = 4 #The code will compute derivatives up to this order
N_grid = 6 #The size of the stencil will be 2*N_grid + 1

#These are the specifications for the grid. The first value is the reference one, and the second is the step
#size in units of the reference values. 

# Example: If "x":[x0, x_percentage], the grid spacing will be x0*x_percentage

expansion_table = {
    "a":[-1, 0.1],
    "b":[1,  0.1]
}

ND_obj.define_setup(N_taylor,N_grid, expansion_table)
```

Now, we compute the function in the total grid defined above. First, without the multiprocessing:

```python
ND_obj.parallel = False
ND_obj.evaluate_in_grid(function_test, x_array=x_array)
```

Now, to take the derivatives:

```python
ND_obj.take_derivative_ALL(verbose=False)
```

And thats it! Lets plot the results:

```python
for this_derivative in list(ND_obj.derivatives.keys()):
    plt.figure(figsize = (4,4), constrained_layout = True)
    plt.title(this_derivative)
    for this_stencil in list(ND_obj.derivatives[this_derivative]):
        try:
            plt.plot(x_array, ND_obj.derivatives[this_derivative][this_stencil], label = this_stencil)
        except:
            pass
    plt.legend()
    plt.show()
```

The code will use all stencil sizes to compute all derivatives. The stencil size usually controls the precision of the derivative: a larger stencil size will lead to a more accurate derivative. The output is a dictionary with all the derivatives:

```python
list(ND_obj.derivatives.keys())
```

Each of of these entries leads to another dictionary

```python
list(ND_obj.derivatives['d1a'].keys())
```

Those inner dictionaries represent the precision (or stencil size) used to compute the derivative associated with the first key.

# 2nd example - Unit test

Consider the function, f(x,b), defined as:

$$
f(x,b) = x*b^{5}/5!
$$

```python
x_array = np.arange(-10,10,0.01)
def f_UnitTest_1(b,kwargs):
    x_array = kwargs["x_array"]
    x_reshaped = np.tile(x_array, (len(b),1))
    return x_reshaped*np.power(b,5)[:,np.newaxis]/np.math.factorial(5)
```

Now, let us take the first 6 derivative of f with respect to b evaluated at b = 0. They all should be zero, except when for $d^5f/db^5$, which should be equal to $f(x,b) = x$.

```python
ND = nd.FiniteDerivative()
N_taylor = 10 #The code will compute derivatives up to this order
N_grid =  6#The size of the stencil will be 2*N_grid + 1

#These are the specifications for the grid. The first value is the reference one, and the second is the step
#size in units of the reference values. 

# Example: If "x":[x0, x_percentage], the grid spacing will be x0*x_percentage

expansion_table = {
    "b":[0, 0.1],
}

ND.define_setup(N_taylor,N_grid, expansion_table)
```

```python
ND.evaluate_in_grid(f_UnitTest_1, x_array = x_array)
ND.take_derivative_ALL()
```

```python
for this_derivative in list(ND.derivatives.keys()):
    plt.figure(figsize = (3,3), constrained_layout = True)
    plt.title(this_derivative)
    for this_stencil in list(ND.derivatives[this_derivative]):
        try:
            plt.plot(x_array, ND.derivatives[this_derivative][this_stencil].T, label = this_stencil)
        except:
            pass
    plt.legend()
    plt.xlim((-1,1))
    plt.ylim((-1,1))
    plt.show()
```

# 3rd example - Unit test

Now, let us take the derivatives of the sine function:
$$
f(x,a,b) = \cos (x \, a)\sin (x\,b)\;.
$$
The result of taking the derivative $N$ times with respect to $b$ is:
$$
\frac{d^{N+M}f(x,b)}{db^N da^M} = x^{M+N}\, \sin\left(\frac{N\pi}{2}+bx\right) \cos\left(\frac{M\pi}{2}+ax\right).
$$

```python
x_array = np.arange(-10,10,0.01)
def f_UnitTest_2(theta,kwargs):
    a,b = theta
    x_array = kwargs["x_array"]
    return np.sin(x_array*b)*np.cos(x_array*a)
```

```python
def UnitTest_Target(x_array, M, N, a, b):
    return x_array**(N+M)*np.sin(N*np.pi/2 + b*x_array)*np.cos(M*np.pi/2 + a*x_array)
```

```python
ND = nd.FiniteDerivative()
N = 5 #The code will compute derivatives up to this order
N_grid =  10 #The size of the stencil will be 2*N_grid + 1

#These are the specifications for the grid. The first value is the reference one, and the second is the step
#size in units of the reference values. 

# Example: If "x":[x0, x_percentage], the grid spacing will be x0*x_percentage

expansion_table = {
    "a":[1, 0.05],
    "b":[1, 0.05],
}

ND.define_setup(N,N_grid, expansion_table)
ND.evaluate_in_grid(f_UnitTest_2, x_array=x_array,save_file = "output/Unit2.npz")
ND.take_derivative_ALL(save_file = "output/Unit2.npz")
```

```python
import re
def extract_numbers(s):
    # find the number before 'a' or 0
    match = re.search(r'\d+(?=a|$)', s)
    a_number = int(match.group()) if match else 0
    
    # find the number before 'b' or 0
    match = re.search(r'\d+(?=b|$)', s)
    b_number = int(match.group()) if match else 0
    
    return a_number, b_number
```

```python
for this_derivative in list(ND.derivatives.keys()):
    this_M, this_N = extract_numbers(this_derivative)
    fig, ax = plt.subplots(2,1,figsize = (8,8), constrained_layout = True, gridspec_kw={'height_ratios': [3,1]},sharex=True)
    unit_test_target =  UnitTest_Target(x_array, this_M, this_N, 1,1)
    ax[0].set_title(this_derivative)
    ax[0].plot(x_array, unit_test_target, color = "green", lw = 10, label = "Target", alpha = 0.5)
    for this_stencil in list(ND.derivatives[this_derivative]):
        try:
            ax[0].plot(x_array, ND.derivatives[this_derivative][this_stencil].T, label = this_stencil)
            ax[1].plot(x_array, (ND.derivatives[this_derivative][this_stencil].T-unit_test_target)/unit_test_target, label = this_stencil)
        except:
            pass
    ax[0].legend()
    ax[1].set_ylabel(r"$\Delta$", size = 30)
    ax[1].set_ylim((-0.01,0.01))
    ax[0].set_xlim((-10,10))
    ax[1].set_xlim((-10,10))
    plt.show()
```

# 4th example - Cosmology: linear matter power spectrum

I will use the example in the CLASS-PT website to get the linear matter power-spectrum, which is obtained by solving some complicated equations numerically. A single iteration of the CLASS-PT will take a some seconds, so here I will show how you can use the multiprocessing library to evaluate the function in the grid. 

The starting point is to define the function in a separate Python file and import it along with other necessary values.

```python
from my_functions import ComputePk,kvec
```

The function should take as input $\omega_{\rm cdm}$ and $h$:

```python
theta = [0.119, 0.679] #omega_cdm, h
PK = ComputePk(theta)
```

```python
plt.figure(figsize = (6,4), constrained_layout= True)
plt.plot(kvec, kvec*PK)
plt.xlim((kvec[0],0.2))
plt.xlabel(r"$k$ [h/Mpc]", size = 20)
plt.ylabel(r"$k\times P(k)$ [Mpc/$h$]$^2$", size = 20)
```

Now, let's see how we can compute the derivative of the linear matter power spectrum with respect to $\omega_{\rm cdm}$ and $h$. I will do it two ways. The first one, I won't use multiprocessing.

```python
ND_obj = nd.FiniteDerivative()

N_taylor = 3 #The code will compute derivatives up to this order
N_grid = 3 #The size of the stencil will be 2*N_grid + 1

expansion_table = {
    "omega_cdm":[0.1190, 0.1190*0.1],
    "h":[0.678,0.678*0.1]
}

ND_obj.define_setup(N_taylor,N_grid, expansion_table)
```

```python
start = time.time()
ND_obj.parallel = False
ND_obj.evaluate_in_grid(ComputePk)
end = time.time()
print("Time to compute the Pk in the grid without using the multiprocessing: {}".format(end-start))
```

```python
start = time.time()
ND_obj.parallel = True
ND_obj.processes = 8
ND_obj.evaluate_in_grid(ComputePk, save_file= "output/PK_grid.npz")
end = time.time()
print("Time to compute the Pk in the grid using the multiprocessing: {}".format(end-start))
```

```python
ND_obj.take_derivative_ALL()
```

```python
for this_derivative in list(ND_obj.derivatives.keys()):
    plt.figure(figsize = (8,8), constrained_layout = True)
    plt.title(this_derivative)
    for this_stencil in list(ND_obj.derivatives[this_derivative]):
        try:
            plt.plot(kvec, kvec*ND_obj.derivatives[this_derivative][this_stencil], label = this_stencil)
            plt.xlim((0.005,0.2))
        except:
            pass
    plt.legend()
    plt.show()
```
